```python
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
MY_TABLE_STYLES = [
    {'selector': 'th', 'props': [
            ('background-color', 'lightgrey'),
            ('font-size', '9px'),
            ('font-family', 'sans-serif'),
            ('text-align', 'left'),
            ('word-wrap', 'break-word'),
            ('padding', '6px 10px')
        ]},
        # Data cells: white background, smaller font, tight padding, NO WORD WRAP
        {'selector': 'td', 'props': [
            ('background-color', 'white'),
            ('font-size', '9px'),
            ('font-family', 'sans-serif'),
            ('white-space', 'nowrap'),   # <-- Prevents ugly text wrapping
            ('padding', '4px 10px')      # <-- Tighter padding makes rows shorter
        ]},
        # Alternating row colors (Zebra Striping / nth-child)
        {'selector': 'tr:nth-child(even) td', 'props': [
            ('background-color', '#f9f9f9')
        ]},
        # Hover effect
        {'selector': 'tr:hover td', 'props': [
            ('background-color', 'lightblue')
        ]}
]

def show_table(df, rows=5, title=None):
    if title: print(f"=== {title.upper()} ===")

    preview = df.head(rows)

    num_cols = preview.select_dtypes(include=np.number).columns

    return (
        preview.style
        .set_table_styles(MY_TABLE_STYLES)
        .hide(axis="index")
        .set_properties(subset=num_cols, **{'text-align': 'right'})
        .highlight_null(color='rgba(255, 0, 0, 0.3)')
        .format(precision=2)
    )

def show_schema(df):
    schema = (
        df.dtypes
        .rename("dtype")
        .to_frame()
        .reset_index(names="column")
    )

    return (
        schema.style
        .set_table_styles(MY_TABLE_STYLES)
        .hide(axis="index")
    )
```

```python
RAW_DIR = Path("../data/raw/pcdd/2026-06")

paths = {
    "rate": RAW_DIR / "pcdem-nhs-rate-jun-2026.csv",
    "sub_icb": RAW_DIR / "pcdem-sub-icb-jun-2026.csv",
    "practice": RAW_DIR / "pcdem-practice-jun-2026.csv",
    "mapping": RAW_DIR / "mapping-file-july-dementia-2026.csv",
    "dictionary": RAW_DIR / "PCDD-2627-data-dictionary.xlsx",

}

rate = pd.read_csv(paths["rate"])
sub_icb = pd.read_csv(paths["sub_icb"])
practice = pd.read_csv(paths["practice"])
mapping = pd.read_csv(paths["mapping"])
dictionary = pd.read_excel(paths["dictionary"])

datasets = {"rate": rate, "sub_icb": sub_icb, "practice": practice, "mapping": mapping}

for name, dataset in datasets.items():
    print(f'Displaying dataset: {name}')
    display(show_table(dataset))
    display(show_schema(dataset))
    print(f'shape of {name}: {dataset.shape}')
    print()
```
