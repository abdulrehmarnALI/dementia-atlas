```python
MY_TABLE_STYLES = [
    {
        "selector": "th",
        "props": [
            ("background-color", "lightgrey"),
            ("font-size", "9px"),
            ("font-family", "sans-serif"),
            ("text-align", "left"),
            ("word-wrap", "break-word"),
            ("padding", "6px 10px"),
        ],
    },
    # Data cells: white background, smaller font, tight padding, NO WORD WRAP
    {
        "selector": "td",
        "props": [
            ("background-color", "white"),
            ("font-size", "9px"),
            ("font-family", "sans-serif"),
            ("white-space", "nowrap"),  # <-- Prevents ugly text wrapping
            ("padding", "4px 10px"),  # <-- Tighter padding makes rows shorter
        ],
    },
    # Alternating row colors (Zebra Striping / nth-child)
    {"selector": "tr:nth-child(even) td", "props": [("background-color", "#f9f9f9")]},
    # Hover effect
    {"selector": "tr:hover td", "props": [("background-color", "lightblue")]},
]


def show_table(df, rows=5, title=None):
    if title:
        print(f"=== {title.upper()} ===")

    preview = df.head(rows)

    num_cols = preview.select_dtypes(include=np.number).columns

    return (
        preview.style.set_table_styles(MY_TABLE_STYLES)
        .hide(axis="index")
        .set_properties(subset=num_cols, **{"text-align": "right"})
        .highlight_null(color="rgba(255, 0, 0, 0.3)")
        .format(precision=2)
    )


def show_schema(df):
    schema = df.dtypes.rename("dtype").to_frame().reset_index(names="column")

    return schema.style.set_table_styles(MY_TABLE_STYLES).hide(axis="index")
```

## Rate Dataset

**Publisher definition:** Diagnosis rate indicator by sub ICB location - ICB - region - England

**Shape:** 750 rows × 9 columns

### What does one row represent?

The rate dataset reports dementia measures for NHS organisations/georgaphies at the reporting period end date.

Key fields:
- `NAME` - organisation name
    - an organisation is a boundary definition such as 'ENGLAND' or 'NHS NORTH WEST YORKSHIRE ICB - X2C4Y
- `ACH_DATE` - achieved date
    - the achieved date represnts the reporting period end date
    - this terminology is inherited from the Quality and Outcomes Framework (QOF) system
- `MEASURE` - identifies what the `VALUE` represents
- `VALUE` - alue associated with the measure
- `ORG_TYPE` - organisation breakdown (geographic level)
- `ORG_CODE` - ODS (Organisation Data Service) organisation identifier
- `ONS_CODE` - 9-digit geography code of the organisation
- `DQ` - "Identifies organisations with a smaller denominator population than the CFAS II reference population with a ‘1’. Values for these organisations should be interpreted with caution."

Key measures:
|Measure|Meaning|
|---|---|
| DEMENTIA_REGISTER_65_PLUS | Number of patients aged 65 and over with a recorded dementia diagnosis |
| DEMENTIA_ESTIMATE_65_PLUS | Expected number of patients aged 65 and over with dementia |
| DIAG_RATE_65_PLUS | Estimated dementia diagnosis rate (recorded/expected) in patients aged 65 and over |
| DIAG_RATE_65_PLUS_LL | Lower 95% confidence limit of estimated dementia diagnosis rate in patients aged 65 and over |
| DIAG_RATE_65_PLUS_UL | Upper 95% confidence limit of estimated dementia diagnosis rate in patients aged 65 and over |

## Sub-ICB Dataset

**Publisher definition:** Primary care dementia data: Dementia data by sub ICB location

### What does one row represent?

The Sub-ICB dataset reports a measure/value for a Sub-ICB at the reporting period end date, with some measures further broken down by dimensions such as age, gender, ethnicity, dementia type or residential type.

Key fields:

- `ACH_DATE` - achieved date
    - represents the reporting period end date
    - this terminology is inherited from the Quality and Outcomes Framework (QOF) system

- `MEASURE` - identifies what the `VALUE` represents

- `BREAKDOWN` - identifies whether/how the measure has been broken down into a more specific group
    - e.g. `AGE_GENDER` indicates that the measure is split by age and gender

- `AGE` - age group for the observation
    - `ALL` indicates the measure is not restricted to a particular age group

- `GENDER` - gender group for the observation
    - `ALL` indicates the measure is not restricted to a particular gender

- `ETHNICITY` - ethnicity group for the observation
    - `ALL` indicates the measure is not restricted to a particular ethnicity group

- `DEMENTIA_TYPE` - dementia type for the observation
    - `ALL` indicates the measure is not restricted to a particular dementia type

- `RESIDENTIAL_TYPE` - residential setting/type for the observation
    - `ALL` indicates the measure is not restricted to a particular residential type

- `VALUE` - value associated with the measure and its applicable breakdown
