## Rate Dataset

**Publisher definition:** Diagnosis rate indicator by sub ICB location - ICB - region - England

**Shape:** 750 rows × 9 columns

### What does one row represent?

The rate dataset reports dementia measures for NHS organisations/georgaphies at the reporting period end date.

Key fields:
- `NAME` - organisation name
    - an organisation is a boundary definition such as 'ENGLAND' or 'NHS NORTH WEST YORKSHIRE ICB - X2C4Y
- `ACH_DATE` - achieved date
    - the achieved date represnts the reporting period end date
    - this terminology is inherited from the Quality and Outcomes Framework (QOF) system
- `MEASURE` - identifies what the `VALUE` represents
- `VALUE` - alue associated with the measure
- `ORG_TYPE` - organisation breakdown (geographic level)
- `ORG_CODE` - ODS (Organisation Data Service) organisation identifier
- `ONS_CODE` - 9-digit geography code of the organisation
- `DQ` - "Identifies organisations with a smaller denominator population than the CFAS II reference population with a ‘1’. Values for these organisations should be interpreted with caution."

Key measures:
|Measure|Meaning|
|---|---|
| DEMENTIA_REGISTER_65_PLUS | Number of patients aged 65 and over with a recorded dementia diagnosis |
| DEMENTIA_ESTIMATE_65_PLUS | Expected number of patients aged 65 and over with dementia |
| DIAG_RATE_65_PLUS | Estimated dementia diagnosis rate (recorded/expected) in patients aged 65 and over |
| DIAG_RATE_65_PLUS_LL | Lower 95% confidence limit of estimated dementia diagnosis rate in patients aged 65 and over |
| DIAG_RATE_65_PLUS_UL | Upper 95% confidence limit of estimated dementia diagnosis rate in patients aged 65 and over |

## Sub-ICB Dataset

**Publisher definition:** Primary care dementia data: Dementia data by sub ICB location

**Shape:** 11,660 rows × 13 columns

### What does one row represent?

The Sub-ICB dataset reports one measure/value for a Sub-ICB at the reporting period end date.

Some measures are further split into breakdowns such as age and gender, ethnicity, dementia type, residential type, prescribing, frailty or referrals.

The effective grain is therefore:

`Sub-ICB + reporting date + measure + applicable breakdown/category = one value`

Key fields:

- `ACH_DATE` - achieved date
    - represents the reporting period end date
    - this terminology is inherited from the Quality and Outcomes Framework (QOF) system
- `MEASURE` - identifies the main type of measure being reported
- `BREAKDOWN` - identifies the specific breakdown applied to the measure
    - e.g. `AGE_GENDER`, `ETHNICITY`, `DEMENTIA_TYPE`
    - some measures have no additional breakdown and contain `NaN`
- `AGE` - age category for the observation
    - `ALL` indicates the value is not restricted to a particular age category
- `GENDER` - gender category for the observation
    - `ALL` indicates the value is not restricted to a particular gender
- `ETHNICITY` - ethnicity category for the observation
    - `ALL` indicates the value is not restricted to a particular ethnicity category
- `DEMENTIA_TYPE` - dementia type category for the observation
    - `ALL` indicates the value is not restricted to a particular dementia type
- `RESIDENTIAL_TYPE` - residential category for the observation
    - `ALL` indicates the value is not restricted to a particular residential type
- `VALUE` - count of the measure for the applicable breakdown

## Practice Dataset

**Publisher definition:** Primary care dementia data: Recorded diagnoses, care plans, reviews by GP practice

**Shape:** 42,616 rows × 7 columns

### What does one row represent?

The practice dataset reports one measure/breakdown value for a GP practice at the reporting period end date.

The effective grain is:

`GP practice + reporting date + measure + breakdown = one value`

Key fields:

- `ACH_DATE` - reporting period end date
- `ORG_TYPE` - organisational breakdown
    - all rows in this file use `PRACTICE`
- `ODS_CODE` - Organisation Data Service identifier for the GP practice
- `NAME` - name of the GP practice
- `MEASURE` - groups related data items into a top-level measure
- `BREAKDOWN` - identifies the specific data item within the measure
- `VALUE` - count of the measure/breakdown

### Practice Dataset Structure Breakdown

The dataset contains **42,616 observations** across **6,088 GP practice codes**.

Every practice contributes exactly **7 rows** in this release:

| Measure | Rows per Practice | Breakdown |
| :--- | ---: | :--- |
| `DEMENTIA_REGISTER` | 2 | aged 0-64; aged 65+ |
| `PAT_LIST` | 2 | aged 0-64; aged 65+ |

## Mapping Dataset

**Publisher definition:** Primary care dementia data: current month: Mapping (Commissioning Regions-ICBs-SICBLs-GP practice)

**Shape:** 6,182 rows × 13 columns

### What does one row represent?

The mapping dataset is a snapshot linking one GP practice to the NHS geographies/organisations above it at the extraction date.

The effective grain is:

`GP practice at extract date = one mapping row`

Key fields:

- `EXTRACT_DATE` - date the mapping snapshot was extracted
- `PRACTICE_CODE` - GP practice code
- `PRACTICE_NAME` - GP practice name
- `PRACTICE_POSTCODE` - GP practice postcode
- `ONS_SUB_ICB_LOCATION_CODE` - ONS 9-digit geography code for the Sub-ICB location
- `SUB_ICB_LOCATION_CODE` - ODS code for the Sub-ICB location
- `SUB_ICB_LOCATION_NAME` - Sub-ICB location name
- `ONS_ICB_CODE` - ONS 9-digit geography code for the ICB
- `ICB_CODE` - ODS code for the ICB
- `ICB_NAME` - ICB name
- `ONS_COMM_REGION_CODE` - ONS 9-digit geography code for the NHS commissioning region
- `COMM_REGION_CODE` - ODS code for the NHS commissioning region
- `COMM_REGION_NAME` - NHS commissioning region name


### PRACTICE / GEOGRAPHY COVERAGE

### PRACTICE DATASET COMPARISON

```python
# Compare GP practice coverage between the practice observations and mapping snapshot.
practice_codes = set(practice["ODS_CODE"])
mapping_codes = set(mapping["PRACTICE_CODE"])

print(f"Practice dataset codes: {len(practice_codes):,}")
print(f"Mapping dataset codes: {len(mapping_codes):,}")
print(f"Present in both: {len(practice_codes & mapping_codes):,}")
print(f"Mapping only: {len(mapping_codes - practice_codes):,}")
print(f"Practice data only: {len(practice_codes - mapping_codes):,}")
```

### Mapping Dataset Structure Breakdown

The mapping file contains **6,182 rows** and **6,182 unique GP practice codes**, confirming one mapping row per practice code in this snapshot.

The hierarchy represented is:

`GP practice → Sub-ICB location → ICB → NHS commissioning region`

The mapping contains:

- **106** Sub-ICB location codes
- **36** ICB codes
- **7** NHS commissioning region codes

There is **1 practice row** with missing Sub-ICB, ICB and commissioning-region mapping fields.

Comparing the mapping snapshot with the practice observations:

- all **6,088** practice codes in the practice dataset are present in the mapping file
- the mapping file contains an additional **94** practice codes that do not appear in the June practice observations

The current files establish that the mapping snapshot covers more practice codes than the practice dataset; the reason for those additional 94 practices is not established at this stage.
